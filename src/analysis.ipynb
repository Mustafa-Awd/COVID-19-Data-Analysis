#### Load Data

```python
from pathlib import Path

BASE_DIR = Path().cwd().parent
DATA_PATH = BASE_DIR / "data" / "covid_worldwide.csv"
```

```python
import pandas as pd

df = pd.read_csv(DATA_PATH)
df.head(3)
```

#### Inspecting Data

```python
print("(Mean deaths, Mean recovered)")
int(df["Total Deaths"].mean()), int(df["Total Recovered"].mean())
```

```python
columns_with_missing_values = df.columns[df.isna().any()].tolist()
print(f"Columns with missing data: {columns_with_missing_values}")
```

```python
missing_values_per_column = []

for column in columns_with_missing_values:
    missing_values_per_column.append(len(df[df[column].isnull()]))

missing_values_per_column
```
